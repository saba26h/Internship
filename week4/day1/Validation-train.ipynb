# Train, Validation, and Test Splits with Model Evaluation

#### Lab Overview:

In this laboratory session, the dataset is prepared and divided into three subsets: Training, Validation, and Testing datasets. This approach allows machine learning models to be trained, tuned, and evaluated in a reliable and unbiased manner.

The workflow covered in this notebook includes:

- Data preparation and preprocessing
- Train / Validation / Test splitting
- Logistic Regression classification
- Random Forest classification
- Hyperparameter tuning using the validation set
- Final evaluation on the test set

Two classification algorithms are trained using the same training dataset and evaluated using the same validation and testing datasets to ensure a fair comparison.

The validation set is used to select the best model configurations, while the test set is reserved for the final evaluation. Model performance is measured using **Accuracy**, **F1-Score**, and **Classification Reports** to determine which classifier performs best on this dataset.

## Load and Prepare the Dataset

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, f1_score
from sklearn.linear_model import LogisticRegression

df=pd.read_csv('students_train.csv')
x=df.drop(['Pass', 'Student_ID'], axis=1)
y=df['Pass']
```

#### Overview:

Before training any machine learning model, the dataset must be loaded and prepared. Proper data preparation ensures that the models receive clean and relevant input features, which improves the quality of the predictions.

In this step, the required libraries for data manipulation, model training, and evaluation are imported. The dataset is then loaded using Pandas from a CSV file.

The target variable (**Pass**) is separated from the input features because it represents the value that the models will learn to predict. The **Student_ID** column is removed since it is only an identifier and does not provide meaningful information for classification.

At the end of this step, the dataset is organized into:

- **X**: Input features used for training the models.
- **y**: Target labels representing whether a student passed or failed.

---

| Function | Description |

| `drop()` | Removes unnecessary columns from the dataset. |

---

- Loads the dataset into the notebook.
- Separates features from the target variable.
- Removes irrelevant information.
- Prepares the data for model training and evaluation.

## Split the Dataset into Training, Validation, and Test Sets

```python
x_temp, x_test, y_temp, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

x_train, x_val, y_train, y_val = train_test_split(x_temp, y_temp, test_size=0.25, random_state=42)
```

#### Overview:

After loading the dataset, it is important to divide the data into separate subsets for training, validation, and testing.

The dataset is first split into a temporary dataset and a testing dataset. The testing dataset is kept completely unseen during model development and is only used for the final evaluation.

The temporary dataset is then divided again into training and validation datasets. The training set is used to train the machine learning models, while the validation set is used to compare models and tune hyperparameters before testing.

This approach ensures that the final performance measurements are reliable and unbiased.

The dataset was successfully split into **60% training, 20% validation, and 20% testing data**.

---

| Function | Description |

---

- Prevents overfitting.

- Provides a fair estimate of model performance.
- Keeps test data completely unseen.

## Train and Evaluate the Logistic Regression Model

```python
tree=LogisticRegression(random_state=42)
tree.fit(x_train, y_train)

pre=tree.predict(x_val)
print("Predictions:\n", pre)
print("Accuracy:", accuracy_score(y_val, pre))
print("F1 Score:", f1_score(y_val, pre))
print("Classification Report:\n", classification_report(y_val, pre))
```

#### Overview:

In this step, a Logistic Regression model is created and trained using the training dataset.

After the training process is completed, the model generates predictions for the validation dataset. The predicted values are then compared with the actual labels to measure the model's performance.

Several evaluation metrics are calculated, including Accuracy, F1 Score, and a Classification Report. These metrics help determine how effectively the model can classify students as passing or failing.

## Train and Evaluate the Random Forest Model

##

```python
forest = RandomForestClassifier(n_estimators=100, random_state=42)
forest.fit(x_train, y_train)
forest_pred = forest.predict(x_val)
accuracy = accuracy_score(y_val, forest_pred)
f1 = f1_score(y_val, forest_pred)
report = classification_report(y_val, forest_pred)

print("Random Forest Predictions:\n", forest_pred)
print("Random Forest Accuracy:", accuracy)
print("Random Forest F1 Score:", f1)
print("Random Forest Classification Report:\n", report)
```

#### Overview:

A Random Forest classifier is trained using the training dataset.

Random Forest is an ensemble learning algorithm that combines the predictions of multiple decision trees to improve classification performance and reduce overfitting.

After training, predictions are generated for the validation dataset. The model is then evaluated using Accuracy, F1 Score, and a Classification Report to determine its effectiveness.

The obtained results can later be compared with Logistic Regression to identify the better-performing model.

## Hyperparameter Tuning for Logistic Regression

```python
c_values = [0.01, 0.1, 1, 10, 100]
best_c = None
best_accuracy = 0
for c in c_values:
    reg = LogisticRegression(C=c, random_state=42)
    reg.fit(x_train, y_train)
    reg_pred = reg.predict(x_val)
    accuracy = accuracy_score(y_val, reg_pred)
    f1 = f1_score(y_val, reg_pred)
    print(f"Logistic Regression with C={c}:")
    print("Accuracy:", accuracy)
    print("F1 Score:", f1)
    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_c = c
print(f"Best C: {best_c}, Best Accuracy: {best_accuracy}")
```

#### Overview:

Machine learning models contain parameters that can significantly affect performance.

In this step, several values of the regularization parameter **C** are tested. For each value, a Logistic Regression model is trained and evaluated on the validation dataset.

The accuracy and F1 score are recorded for every configuration. The value that achieves the highest validation accuracy is selected as the optimal parameter for the final model.

## Hyperparameter Tuning for Random Forest

```python
n_estimators_values = [10, 50, 100, 200]
best_n_estimators = None
best_accuracy_forest = 0
for n in n_estimators_values:
    forest=RandomForestClassifier(n_estimators=n, random_state=42)
    forest.fit(x_train, y_train)
    forest_pred = forest.predict(x_val)
    accuracy = accuracy_score(y_val, forest_pred)
    f1 = f1_score(y_val, forest_pred)
    print(f"Random Forest with n_estimators={n}:")
    print("Accuracy:", accuracy)
    print("F1 Score:", f1)
    if accuracy > best_accuracy_forest:
        best_accuracy_forest = accuracy
        best_n_estimators = n
print(f"Best n_estimators: {best_n_estimators}, Best Accuracy: {best_accuracy_forest}")
```

#### Overview:

The number of trees used in a Random Forest model can influence both performance and computational cost.

In this step, several values of **n_estimators** are evaluated. For each value, a Random Forest model is trained and tested on the validation dataset.

The Accuracy and F1 Score are calculated for every configuration. The model achieving the highest validation accuracy is selected for final testing.

## Final Evaluation of the Logistic Regression Model

```python
final_model_log=LogisticRegression(C=best_c, random_state=42)
final_model_log.fit(x_train, y_train)
final_log_pred = final_model_log.predict(x_test)
final_log_accuracy = accuracy_score(y_test, final_log_pred)
final_log_f1 = f1_score(y_test, final_log_pred)
final_log_report = classification_report(y_test, final_log_pred)
print("Final Logistic Regression Model:")
print("Accuracy:", final_log_accuracy)
print("F1 Score:", final_log_f1)
print("Classification Report:\n", final_log_report)
```

#### Overview:

After selecting the optimal value of C, the final Logistic Regression model is trained using the chosen configuration.

The trained model is then evaluated on the testing dataset, which has not been used during training or validation.

Accuracy, F1 Score, and a Classification Report are generated to provide an unbiased estimate of the model's real-world performance.

## Final Evaluation of the Random Forest Model

```python
final_model_forest = RandomForestClassifier(n_estimators=best_n_estimators, random_state=42)
final_model_forest.fit(x_train, y_train)
final_forest_pred = final_model_forest.predict(x_test)
final_forest_accuracy = accuracy_score(y_test, final_forest_pred)
final_forest_f1 = f1_score(y_test, final_forest_pred)
final_forest_report = classification_report(y_test, final_forest_pred)
print("Final Random Forest Model:")
print("Accuracy:", final_forest_accuracy)
print("F1 Score:", final_forest_f1)
print("Classification Report:\n", final_forest_report)
```
