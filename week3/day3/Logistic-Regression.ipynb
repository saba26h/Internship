# Logistic Regression Classification

## 1.Building a Logistic Regression Classifier

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

df=pd.read_csv('student_pass.csv')

x=df.drop('Passed',axis=1)
y=df['Passed']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

model=LogisticRegression(max_iter=1000)

model.fit(x_train,y_train)
predictions=model.predict(x_test)
print(predictions)
probabilities=model.predict_proba(x_test)
print(probabilities)
```

#### Explanation:

This code builds a Logistic Regression model to predict whether a student will pass or fail.

**Steps performed:**
- Import the required libraries.
- Load the dataset.
- Separate the features (`X`) from the target (`y`).
- Split the data into training and testing sets.
- Train the Logistic Regression model.
- Predict the class labels for the test data.
- Predict the probability of each class.

#### Interpretation:

- `predictions` contains the predicted class (Pass or Fail).
- `probabilities` contains the confidence score for each prediction.
- Logistic Regression estimates the probability that an observation belongs to the positive class.
- These probabilities are later used to calculate ROC-AUC.

## 2.Evaluating the Logistic Regression Model

```python
from sklearn.metrics import confusion_matrix,classification_report

conf_matrix=confusion_matrix(y_test,predictions)
class_report=classification_report(y_test,predictions)

print("Confusion Matrix:\n",conf_matrix)
print("Classification Report:\n",class_report)
```

#### Explanation:

This section evaluates the model using a confusion matrix and a classification report.

The confusion matrix shows:
- True Positives (TP)
- True Negatives (TN)
- False Positives (FP)
- False Negatives (FN)

The classification report includes:
- Precision
- Recall
- F1-score
- Accuracy

#### Interpretation:

- **Precision** measures how many predicted positive cases are actually positive.
- **Recall** measures how many actual positive cases were correctly identified.
- **F1-score** balances Precision and Recall.
- Higher values indicate better classification performance.

## 3.ROC Curve and AUC Score

```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(y_test, probabilities[:, 1])
print("ROC AUC Score: \n", roc_auc)
```

#### Explanation:

The ROC-AUC score evaluates how well the model distinguishes between the two classes.

- ROC stands for Receiver Operating Characteristic.
- AUC stands for Area Under the Curve.

The score ranges from 0 to 1.

#### Interpretation:

- **AUC = 1.0** → Perfect classifier.
- **AUC ≈ 0.9** → Excellent performance.
- **AUC ≈ 0.8** → Good performance.
- **AUC ≈ 0.7** → Acceptable performance.
- **AUC = 0.5** → No better than random guessing.

A higher ROC-AUC score indicates that the Logistic Regression model is better at distinguishing between students who pass and those who fail.
