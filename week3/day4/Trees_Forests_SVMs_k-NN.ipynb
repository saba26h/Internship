# Trees, Forests, SVMs, and K-Nearest Neighbors

#### Lab Overview:

In this laboratory session, several supervised machine learning algorithms are implemented and compared using the same dataset. The objective is to understand how different classification algorithms learn from data, make predictions, and evaluate their performance.

The algorithms covered in this notebook are:

-  Decision Tree
-  Random Forest
-  Support Vector Machine (SVM)
-  K-Nearest Neighbors (K-NN)

Each algorithm is trained using the same training dataset and evaluated using the same testing dataset to ensure a fair comparison. Finally, the models are compared using the **Weighted F1-Score** to determine which classifier performs best on this problem.

## Step 1: Load and Prepare the Dataset

````python
import pandas as pd
from sklearn.model_selection import train_test_split

df=pd.read_csv('students_train.csv')
x=df.drop(['Pass', 'Student_ID'], axis=1)
y=df['Pass']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print("Actual Values:\n", y_test)
````

#### Overview:

Before building any machine learning model, the dataset must be loaded and prepared. Data preparation is one of the most important stages because the quality of the input data directly affects the quality of the model.

In this step, the dataset is imported using **Pandas**. The target column (**Pass**) is separated from the input features, while the **Student_ID** column is removed because it does not contain useful information for prediction.

The dataset is then divided into training and testing sets using `train_test_split()`. The training data is used to teach the models, while the testing data is reserved for evaluating the models on unseen samples.

---

#### Functions Used:

| Function | Description |
|-----------|-------------|
| `pd.read_csv()` | Loads the dataset from a CSV file. |
| `drop()` | Removes unnecessary columns. |
| `train_test_split()` | Splits the dataset into training and testing sets. |

---

#### Why is this step important?:

- Prevents data leakage.
- Allows fair evaluation.
- Separates learning from testing.
- Improves model reliability.

## Step 2: Decision Tree Classifier

##

````python
from sklearn.tree import DecisionTreeClassifier
tree=DecisionTreeClassifier(max_depth=5, random_state=42)
tree.fit(x_train, y_train)

pre=tree.predict(x_test)
print("Predictions:", pre)
````

#### Overview:

A Decision Tree is one of the most popular supervised learning algorithms. It predicts the target value by creating a sequence of decisions based on the input features.

The algorithm starts from a root node and repeatedly splits the dataset according to the feature that provides the highest information gain. This process continues until the stopping condition is reached, resulting in a tree structure where each leaf node represents the final prediction.

Because every decision can be visualized, Decision Trees are considered one of the easiest machine learning algorithms to interpret.

---

#### Functions Used:

| Function | Purpose |
|-----------|----------|
| `DecisionTreeClassifier()` | Creates the Decision Tree model. |
| `fit()` | Trains the model. |
| `predict()` | Predicts unseen samples. |

---

#### Important Parameters:

| Parameter | Description |
|------------|-------------|
| `max_depth=5` | Limits tree depth to reduce overfitting. |
| `random_state=42` | Ensures reproducible results. |

---

#### Advantages:

- Easy to understand.
- Easy to visualize.
- Fast training.
- Works with numerical and categorical features.

---

#### Limitations:

- Can overfit easily.
- Sensitive to noisy data.
- Small data changes may produce a different tree.

---

#### Key Takeaway:

Decision Trees are excellent when model interpretability is important, but controlling tree depth is necessary to avoid overfitting.

## Step 3: Random Forest Classifier

##

````python
from sklearn.ensemble import RandomForestClassifier
forest=RandomForestClassifier( max_depth=5, random_state=42)
forest.fit(x_train, y_train)
print("Feature Importances:", forest.feature_importances_)

pre=forest.predict(x_test)
print("Predictions:", pre)
````

#### Overview:

Random Forest is an ensemble learning algorithm that combines multiple Decision Trees into one powerful model.

Instead of relying on a single tree, the algorithm builds many trees using random subsets of the training data. Each tree makes its own prediction, and the final result is determined through majority voting.

This strategy significantly improves prediction accuracy while reducing overfitting.

---

#### Functions Used:

| Function | Description |
|-----------|-------------|
| `RandomForestClassifier()` | Creates the Random Forest model. |
| `fit()` | Trains all decision trees. |
| `predict()` | Predicts using majority voting. |
| `feature_importances_` | Displays feature importance values. |

---

#### Decision Tree vs Random Forest:

| Decision Tree | Random Forest |
|---------------|---------------|
| Single tree | Multiple trees |
| Faster | Slightly slower |
| More overfitting | Less overfitting |
| Lower accuracy | Higher accuracy |
| Easier to interpret | Harder to interpret |

---

#### Advantages:

- Higher accuracy.
- More stable predictions.
- Less sensitive to noise.
- Reduces overfitting.

---

#### Limitations:

- Longer training time.
- Requires more memory.
- Harder to interpret than a single tree.

---

#### Key Takeaway:

Random Forest improves Decision Trees by combining multiple models, making it one of the most reliable classification algorithms.

## Step 4: Support Vector Machine (SVM)

##

````python
from sklearn.svm import SVC
svm=SVC(kernel='rbf', probability=True)
svm.fit(x_train, y_train)

pre=svm.predict(x_test)
print("Predictions:", pre)
````

#### Overview:

Support Vector Machine (SVM) is a supervised learning algorithm that separates different classes by finding the optimal decision boundary, known as a hyperplane.

Unlike Decision Trees, SVM attempts to maximize the distance between the separating boundary and the closest training samples (support vectors). This usually leads to better generalization on unseen data.

In this notebook, the Radial Basis Function (RBF) kernel is used, allowing the model to classify data that cannot be separated using a straight line.

---

#### Functions Used:

| Function | Description |
|-----------|-------------|
| `SVC()` | Creates the SVM classifier. |
| `fit()` | Trains the model. |
| `predict()` | Predicts new samples. |

---

#### Parameters:

| Parameter | Description |
|-----------|-------------|
| `kernel='rbf'` | Uses a nonlinear decision boundary. |
| `probability=True` | Enables probability estimation. |

---

#### Decision Tree vs SVM:

| Decision Tree | SVM |
|---------------|-----|
| Tree structure | Hyperplane |
| Easy to explain | Harder to explain |
| Fast | Slower |
| Can overfit | Better generalization |

---

#### Advantages:

- High accuracy.
- Effective for complex datasets.
- Works well in high-dimensional spaces.

---

#### Limitations:

- Computationally expensive.
- Sensitive to parameter tuning.
- Difficult to interpret.

---

#### Key Takeaway:

SVM is one of the strongest classifiers for complex datasets but usually requires more computation than tree-based methods.

## Step 5: K-Nearest Neighbors (K-NN)

##

````python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train, y_train)

pre=knn.predict(x_test)
print("Predictions:", pre)
````

#### Overview:

K-Nearest Neighbors (K-NN) is a simple instance-based learning algorithm. Unlike most machine learning models, K-NN does not build an explicit model during training.

Instead, it stores the training data and classifies new samples by finding the nearest training examples. The predicted class is determined by the majority vote of the nearest neighbors.

---

#### Functions Used:

| Function | Description |
|-----------|-------------|
| `KNeighborsClassifier()` | Creates the KNN model. |
| `fit()` | Stores the training data. |
| `predict()` | Predicts using nearest neighbors. |

---

#### Parameters:

| Parameter | Description |
|-----------|-------------|
| `n_neighbors=5` | Uses the five nearest neighbors. |

---

#### How KNN Works:

```
New Sample

      ●

 ○   ○   ○
      ○
   ○

↓

Find the nearest neighbors

↓

Majority Vote

↓

Prediction
```

---

#### KNN vs SVM:

| KNN | SVM |
|------|------|
| Lazy learner | Model-based learner |
| No actual training | Requires training |
| Slow prediction | Faster prediction |
| Sensitive to K | Sensitive to kernel |

---

#### Advantages:

- Very simple.
- Easy to implement.
- No training complexity.

---

#### Limitations:

- Slow on large datasets.
- Sensitive to irrelevant features.
- Requires feature scaling.

---

#### Key Takeaway:

KNN is an intuitive algorithm that performs well on small datasets but becomes computationally expensive as the dataset size increases.

## Final Model Comparison

##

````python
from sklearn.metrics import f1_score

models = {
    "Decision Tree": tree,
    "Random Forest": forest,
    "SVM": svm,
    "K-NN": knn
}

res = []
for name, model in models.items():
    pre=model.predict(x_test)
    score=f1_score(y_test, pre, average='weighted')
    res.append({
        "Model": name,
        "F1 Score": round(score, 3)
    })
    
    com=pd.DataFrame(res)
    com= com.sort_values(by='F1 Score', ascending=False)
print(com)
````

#### Comparison Table:

| Algorithm | Main Idea | Advantages | Limitations |
|------------|-----------|------------|-------------|
|  Decision Tree | Tree-based decisions | Easy to understand | Overfitting |
|  Random Forest | Multiple decision trees | High accuracy | More computation |
|  SVM | Optimal separating hyperplane | Excellent generalization | Slower training |
|  KNN | Nearest neighbors | Simple implementation | Slow prediction |

---

#### Final Observation:

All four algorithms were trained and tested using the same dataset. Their performances were evaluated using the **Weighted F1-Score**, providing a fair comparison between the models.

Although each algorithm has its own strengths and weaknesses, ensemble methods such as Random Forest generally provide higher prediction accuracy, while Decision Trees remain the easiest models to interpret. SVM is powerful for complex decision boundaries, whereas KNN offers a simple yet effective approach for smaller datasets.

This experiment demonstrates that choosing the best machine learning algorithm depends on the characteristics of the dataset, computational requirements, and the desired balance between accuracy and interpretability.
