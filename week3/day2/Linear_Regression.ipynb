## Step 1: Load the Dataset

```python
import pandas as pd

df=pd.read_csv("student_scores.csv")
df.isnull().sum()
```

#### Explanation:

The dataset was loaded using the pandas library, and the missing values in each column were checked using `isnull().sum()`. This helps identify whether the dataset requires cleaning before it is used for machine learning.

<div align="center">
<hr width="50%">
</div>


## Step 2: Clean the Dataset

```python
df.dropna(inplace=True)
df.isnull().sum()
```

#### Explanation:

Missing values were removed from the dataset using `dropna()`. After cleaning, the missing values were checked again to confirm that the dataset no longer contains incomplete records.

<div align="center">
<hr width="50%">
</div>



## Step 3: Separate Features and Target Variable

```python
x=df.drop("Final_Score", axis=1)
y=df["Final_Score"]
print("The features are: \n", x)
print("The target variable is: \n", y)
```

#### Explanation:

The input features were stored in `x` by removing the `score` column, while the `score` column was stored separately in `y` as the target variable. Printing both variables confirms that the data has been separated correctly.

<div align="center">
<hr width="50%">
</div>



## Step 4: Split the Dataset into Training and Testing Sets

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("Training features: \n", x_train)
print("Testing features: \n", x_test)
print("Training target variable: \n", y_train)
print("Testing target variable: \n", y_test)
```

#### Explanation:

The dataset was split into training and testing sets using an 80/20 ratio with `random_state=42` to ensure reproducible results. The training set is used to train the model, while the testing set is reserved for evaluating its performance on unseen data.
<div align="center">
<hr width="50%">
</div>

## Step 5: Train the Linear Regression Model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

#### Explanation:

A `LinearRegression` model was created and trained using the training data (`x_train` and `y_train`) with the `fit()` method. This allows the model to learn the relationship between the input features and the target variable.
<div align="center">
<hr width="50%">
</div>

## Step 6: Display Model Parameters

```python
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

#### Explanation:

The model coefficients and intercept were displayed. The coefficients show the effect of each feature on the prediction, while the intercept is the baseline value when all features are zero.
<div align="center">
<hr width="50%">
</div>

## Step 7: Evaluate the Model

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np

print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("Root Mean Squared Error:", np.sqrt(mean_squared_error(y_test, y_pred)))
```

#### Explanation:

The trained model was evaluated using MAE, MSE, RMSE, and R² score. These metrics measure prediction error and how well the model fits the test data.
