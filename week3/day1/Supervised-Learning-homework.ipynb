## Step 1: Load the Dataset

```python
import pandas as pd

df=pd.read_csv("homework.csv")
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df.isnull().sum()
```

#### Explanation:

The dataset was loaded using pandas and checked for missing values. Then, all rows containing missing values were removed using `dropna()` to ensure the dataset is clean before training the model.


## Step 1: Separate Features and Target Variable

```python
x=df.drop("Mark", axis=1)
y=df["Mark"]
print("The features are: \n", x)
print("The shape of the features is: ", x.shape)
```

```python
print("The target variable is: \n", y)
print("The shape of the target variable is: ", y.shape)
```

#### Explanation:
The dataset was divided into input features (`x`) and the target variable (`y`). The features contain the input data used for prediction, while the target variable contains the values that the model will learn to predict.

___



## Step 2: Perform an 80/20 train/test split with a fixed random_state.

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=40)
```

#### Explanation:

The dataset was split into training and testing sets using an 80% training ratio and a 20% testing ratio.

___



## Step3: Confirm the shapes of X_train, X_test, y_train, y_test are consistent and print them.

```python
print("Training features: \n", x_train)
print("The shape of the training features is: ", x_train.shape)
```

```python
print("Testing features: \n", x_test)
print("The shape of the testing features is: ", x_test.shape)
```

```python
print("Training target variable: \n", y_train)
print("The shape of the training target variable is: ", y_train.shape)
```

```python
print("Testing target variable: \n", y_test)
print("The shape of the testing target variable is: ", y_test.shape)
```

#### Explanation:

The shapes of the training and testing datasets were printed to verify that the split was performed correctly and that the number of samples in the features matches the corresponding target values.

___



## Step 4: Why the Test Set Should Not Be Used During Training

#### Explanation:

The test set should never be used during training because it is reserved for evaluating the model on unseen data. If the model is trained using the test set, the evaluation will no longer be fair, and the model may appear to perform better than it actually does. Keeping the test set separate helps measure how well the model generalizes to new data.
