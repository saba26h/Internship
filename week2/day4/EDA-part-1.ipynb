## 2.Seaborn: Statistical Visualization & Univariate Analysis

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("students.csv")

sns.histplot(data=df, x="age")
plt.title("Age Histplot")
plt.show()

sns.boxplot(data=df, x="age")
plt.title("Age Boxplot")
plt.show()

sns.countplot(data=df, x="gender")
plt.title("Gender Countplot")
plt.show()

sns.kdeplot(data=df, x="age", fill=True)
plt.title("Age KDE")
plt.show()
```

### Code Explanation

This code imports the required libraries:
- **Seaborn** for data visualization.
- **Matplotlib** for plotting.
- **Pandas** for reading and working with the dataset.

Then, it loads the `students.csv` dataset and creates a histogram of the **age** column to visualize how students' ages are distributed.

## 3.Outlier Detection with the IQR Method

```python
q1=df['age'].quantile(0.25)
q3=df['age'].quantile(0.75)
IQR=q3-q1
outlires=df[(df['age']<q1-1.5*IQR)| (df['age']>q3+1.5*IQR)]
print(outlires)
```

### Code Explanation

This code detects outliers in the **age** column using the Interquartile Range (IQR) method.

- It calculates the first quartile (Q1) and third quartile (Q3).
- It computes the Interquartile Range (IQR).
- It determines the lower and upper limits for acceptable values.
- Finally, it displays all rows containing outlier values.
