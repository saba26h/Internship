### Step 1: Simulating Coin Flips

In this experiment, we simulate flipping a fair coin 10,000 times using NumPy.

A fair coin has two equally likely outcomes:
- Heads (1)
- Tails (0)

According to probability theory, the probability of getting heads is:

**P(Heads) = 0.5**

As the number of flips increases, the observed proportion of heads should become closer to 0.5. This illustrates the **Law of Large Numbers**.

```python
import numpy as np
coins=np.random.choice([0,1],size=10000)

heads=np.sum(coins)

pro=heads/len(coins)
print("Total flips : ",len(coins))
print("Number of heads : ",heads)
print("Proportion of heads : ",pro)
```

#### Interpretation

The proportion of heads is very close to 0.5.

Although each experiment produces slightly different results due to randomness, increasing the number of trials causes the observed probability to approach the theoretical probability.

This demonstrates the **Law of Large Numbers**, an important concept used in probability and machine learning.

### Step 2: Sampling from a Normal Distribution

Many real-world variables approximately follow a **normal distribution**.

The normal distribution is:
- Symmetric
- Bell-shaped
- Centered around its mean

In this experiment we generate random values from a normal distribution using NumPy and visualize them with a histogram.

```python
import matplotlib.pyplot as plt

data=np.random.normal(loc=0,scale=1,size=10000)
plt.hist(data,bins=40)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

#### Interpretation

The histogram forms a bell-shaped curve centered near zero.

Most values are concentrated around the mean, while fewer observations appear farther away from the center.

This confirms the characteristic shape of the normal (Gaussian) distribution, which is widely used in statistics and machine learning.

### Step 3: Conditional Probability

Suppose a class contains:

- 40 males who passed
- 30 females who passed
- 10 males who failed
- 20 females who failed

#### Goal

Find the probability that a student is male given that the student passed:

**P(Male | Passed)**

#### Conditional Probability Formula

P(A|B) = P(A ∩ B) / P(B)


#### Calculation

Number of males who passed = 40

Total number of students who passed = 40 + 30 = 70

Therefore:

P(Male | Passed) = 40 / 70

P(Male | Passed) = 0.5714

P(Male | Passed) = 57.14%

#### Conclusion

The probability that a randomly selected student is male, given that the student passed, is **57.14%**.

Next, we verify this result using a simulation.

```python
import pandas as pd

students=pd.read_csv("students.csv")

passed_students=students[students["passed"]=="Yes"]
total=passed_students['count'].sum()
print("The total of students who pass : ",total)

male_passed=passed_students[passed_students["gender"]=='Male']['count'].sum()
print("The number of males who pass : ",male_passed)

print("P(Male | Passed) = ",male_passed/total)
```

#### Interpretation

The simulation produces the same conditional probability as the manual calculation.

This demonstrates that conditional probability measures the likelihood of an event after additional information is known.

Conditional probability is fundamental in machine learning because predictions are often made based on known features.
