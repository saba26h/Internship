# 1.Measures of Central Tendency

```python
import numpy as np
import pandas as pd

arr=np.array([10, 12, 12, 13, 100])
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Mode:", pd.Series(arr).mode()[0])
```

## Explanation

This code calculates the main measures of central tendency for the given dataset. It determines the average value using the mean, identifies the middle value using the median, and finds the most frequently occurring value using the mode. These measures provide different perspectives on the center of the data and help identify how representative the typical value is, especially when the dataset contains extreme values. 
___


# 2.Measures of Spread & Percentiles and Quartiles

```python
print("Standard Deviation:", np.std(arr))
print("Variance:", np.var(arr))
q1,q3 = np.percentile(arr, [25, 75])
print("Interquartile Range:", q3 - q1)
print("Median in Quartiles:", np.percentile(arr, 50))
```

## Explanation

This code evaluates the variability and distribution of the dataset. The standard deviation and variance measure how spread out the values are around the mean, while the quartiles and interquartile range (IQR) describe the spread of the middle 50% of the data. These statistics help assess data consistency and detect the potential influence of outliers.
