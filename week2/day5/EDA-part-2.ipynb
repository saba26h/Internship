### 1.Import Required Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

#### Explanation:

This code imports the libraries needed for data analysis and visualization.
- **Pandas** (pd): Used for reading and manipulating datasets.
- **Seaborn** (sns): Used for creating statistical visualizations.
- **Matplotlib** (plt): Used for plotting graphs and displaying figures.

___



### 2.Load the Dataset

```python
df=pd.read_csv("students.csv")
```

#### Explanation:

This line reads the dataset from the CSV file and stores it in a DataFrame called df.
- read_csv() loads data from a CSV file.
- df represents the dataset that will be analyzed.

___



### 3.Scatter Plot Visualization

```python
sns.scatterplot(data=df,x='score',y='hours_studied')
plt.show()
sns.boxplot(data=df,x='score',y='hours_studied')
plt.show()
```

#### Explanation:

This scatter plot visualizes the relationship between:
- score (student grades)
- hours_studied (study hours)

Each point represents one student.

**Purpose**:
To determine whether students who study more tend to achieve higher scores.

___



### 4.Box Plot Visualization

```python
corr=df.corr(numeric_only=True)
print(corr)
```

#### Explanation:

This box plot displays the distribution of study hours across different score values.
It helps identify:
- Median
- Quartiles
- Spread of data
- Outliers

**Purpose**:
To understand the distribution of the data and detect unusual values.

___



### 5.Correlation Analysis

```python
sns.heatmap(corr, annot=True, cmap="coolwarm")
```

#### Explanation:

This code calculates the correlation matrix for all numerical columns.
- corr() computes correlation coefficients.
- numeric_only=True ignores non-numeric columns such as gender.

**Purpose**:
To measure the strength of relationships between numerical variables.

**Interpretation**:
- 1 → Strong positive correlation
- 0 → No correlation
- -1 → Strong negative correlation

___



### 6.Correlation Heatmap

```python
sns.pairplot(df, hue="gender")
```

#### Explanation:

This heatmap visualizes the correlation matrix using colors.
Parameters:
- corr → Correlation matrix
- annot=True → Displays correlation values
- cmap="coolwarm" → Uses a blue-to-red color scale

**Purpose**:
To quickly identify strong and weak relationships between variables.
