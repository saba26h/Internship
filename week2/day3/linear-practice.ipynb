# Linear Algebra for ML

### 1.Vectors

```python
import numpy as np

arr=np.array([25,5000,3])
print(arr)
```

#### Example 1: Creating a Vector

A vector is a one-dimensional array that represents a single data sample or a collection of related values.

In Machine Learning, each feature of a sample is stored as one element in the vector.

In this example, the vector represents one customer with three features:

- Age
- Income
- Years as a customer (Tenure)

### 2.Matrices

```python
x=np.array([[25, 50000, 3],
            [40, 80000, 10],
            [33, 62000, 5]])

print(x.shape)
```

#### Example 2: Creating a Matrix

A matrix is a two-dimensional array that stores multiple samples.

Each row represents one sample.

Each column represents one feature.

The shape of a matrix is written as:

(rows, columns)

which means:

(samples, features)

### 3.The Dot Product

```python
w=np.array([25, 50000, 3])
a=np.array([0.1, 0.0002, 1.5])
p=np.dot(a,w)
print(p)
```

#### Example 3: Dot Product

The dot product multiplies corresponding elements of two vectors and sums the results.

This operation is fundamental in Machine Learning because linear models compute predictions using the dot product.

Prediction = Features · Weights

### 4.Matrix Multiplication

```python
n=np.array([[25, 50000, 3], [40, 80000, 10]])
m=np.array([0.1, 0.0002, 1.5])

predictions=n@m

print(predictions)
```

#### Example 4: Matrix Multiplication

Instead of predicting one sample at a time, matrix multiplication allows us to predict multiple samples simultaneously.

The @ operator performs matrix multiplication in NumPy.

Each row of the matrix is multiplied by the weight vector to produce one prediction.

**Shape Matching Rule**

For matrix multiplication to work, the number of columns in the first matrix must equal the number of elements in the second vector.

(m × n) @ (n,) → (m,)

If the inner dimensions do not match, NumPy raises a shape mismatch error.
