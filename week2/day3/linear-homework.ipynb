# Step 1 — Create a Dataset Matrixx

```python
import numpy as np
x= np.array([
    [22, 4500, 2],
    [35, 7200, 8],
    [28, 5600, 4]
])

print("The matrix :\n",x)
print("The shape of matrix is :",x.shape)
```

Create a NumPy matrix containing three samples.

Each row represents one customer.

Each column represents one feature:

- Age
- Income
- Years of experience

# Step 2 — Dot Product

```python
sample = x[0]

w = np.array([0.2, 0.003, 1.2])

manual = (sample[0]*w[0] +sample[1]*w[1] +sample[2]*w[2])

dot_numpy = np.dot(sample, w)

print("Manual:", manual)
print("NumPy:", dot_numpy)
```

# Step 4 — Shape Mismatch Error

```python
wrong_w = np.array([0.2, 0.0003])

x @ wrong_w
```

### Step 4: Understanding a Shape Mismatch Error

The weight vector below contains only two values, while each sample has three features.

Since the number of features does not match the number of weights, matrix multiplication cannot be performed.

To fix the error, the weight vector must contain one weight for every feature in the dataset.
