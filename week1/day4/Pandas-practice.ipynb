1.Loading and Inspecting Data

```python
import pandas as pd
df = pd.read_csv("data.csv")
df.head() 
df.shape 
df.info() 
df.describe()
```

I loaded the dataset and explored its contents. I checked the dataset structure, viewed the first rows, and examined the information and statistics of the data.

2.Selecting and Filtering

```python
df['Duration'] 
df[['Duration', 'Pulse']]
df[df['Duration'] > 60]
df.loc[df["Duration"] > 30, "Pulse"]
```

I selected specific columns and filtered rows based on conditions. This makes it easier to work with only the data that I need.

3.Cleaning Data

```python
df.isnull().sum()
df["Calories"].fillna(df["Calories"].mean(), inplace=True)
df.drop_duplicates(inplace=True)
df.describe()
```

I checked for missing values and handled them using the mean value. I also removed duplicate rows to improve the quality of the dataset.

4.Grouping and Aggregation

```python
df.groupby("Duration")["Pulse"].mean()
df.groupby("Duration").agg({"Pulse": "mean", "Calories": "max"})
```

I grouped the data by duration and calculated summary statistics. This helped me understand patterns and compare values between groups.
