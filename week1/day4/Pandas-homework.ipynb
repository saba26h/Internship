1.Loading and Inspecting Data

```python
import pandas as pd
df = pd.read_csv("abc.csv") 

print(df.shape)
print(df.columns)
print(df.dtypes)
```

I loaded the dataset using Pandas and explored its structure. I checked the number of rows and columns, viewed the column names, and identified the data type of each column.

2.Cleaning Data

```python
print(df.isnull().sum())
print(df["history"].fillna(df["history"].mean(), inplace=True))
```

I checked for missing values in the dataset. I also filled the missing values in the history column using the average value to make the data more complete.

3.Selecting and Filtering

```python
filter = df["language"] >= 50
print(df[filter])
```

I filtered the data to display only the rows where the language score is greater than or equal to 50. This helps focus on specific records that meet a condition.

4.Grouping and Aggregation

```python
df.groupby("language").agg({"maths": "mean", "science": "max"})
```

I grouped the data by language and calculated statistics. I found the average maths score and the maximum science score for each language group.
