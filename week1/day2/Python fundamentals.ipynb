```python
def normalize(value, min_value, max_value):
    return (value - min_value) / (max_value - min_value)
print(normalize(75, 0, 100))
```

```python
class DataPoint:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def distance_from_origin(self):
        return (self.x**2+self.y**2)**0.5

p=DataPoint(3,4)
print(p.distance_from_origin())
```

```python
def mean_min_max(numbers):
    sum=0
    for i in numbers:
        sum+=i
    mean=sum/len(numbers)
    max=numbers[0]
    for i in numbers:
        if i>max:
            max=i
    min=numbers[0]
    for i in numbers:
        if i<min:
            min=i
    dict1={"mean":mean,"min":min,"max":max}
    
    return dict1
try:
    num=[10,20,30,40,50]
    dict1=mean_min_max(num)
    print(dict1)
except Exception as e:
    print(f"Error: {e}")
```

first homework 
this function takes a list of numbers and return a dictionary containing the mean , min and max
using dictionary makes the results easy to access by the key
using try and except to avoid division by zero if the list is empty

```python
class Triangle:
    def __init__(self,base,height):
        self.base=base
        self.height=height
    def area(self):
        return 0.5*self.base*self.height

t=Triangle(10,6)
print(t.area())
```

this class represents a triangle with 2 attributes : base and height
includes a method called 'area' that calculates the area of the triangle using the formula:area = 0.5*base*height
