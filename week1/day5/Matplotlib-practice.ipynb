# 1.The Matplotlib Basics & The Four Core Plot Types

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [10,20,30,40,20,30]

plt.plot(x, y)
plt.xlabel('Time (s)')
plt.ylabel('Value') 
plt.title('Value over Time')
plt.show()

```

# Line Plot Analysis

This line chart shows how the values change over time. The trend increases steadily from the first observation until it reaches its highest value at time 4. After that, the values decrease and then rise again slightly.

The visualization helps identify trends, growth patterns, and fluctuations in the data. It clearly shows that the data does not remain constant and experiences both increases and decreases throughout the observed time period.

```python
plt.scatter(x, y)
plt.xlabel('Time (s)')
plt.ylabel('Value')
plt.title('Scatter Plot of Value over Time')
plt.show()
```

## Scatter Plot Analysis

The scatter plot displays the relationship between time and value using individual points. Each point represents a single observation in the dataset.

This visualization makes it easier to identify clusters, gaps, and potential patterns between variables. The points suggest that larger values generally occur during the middle time periods, while lower values are found at the beginning and end of the dataset.


```python
plt.bar(x, y) 
plt.xlabel('Time (s)')
plt.ylabel('Value')
plt.title('Bar Chart of Value over Time')
plt.show()
```

## Bar Chart Analysis

The bar chart compares values across different time points. Unlike the line chart, it emphasizes the magnitude of each observation separately.

The chart shows that the highest value occurs at time 4, while the lowest value appears at time 1. This makes it easy to compare individual observations and identify which categories contribute the most or least to the dataset.

```python
plt.hist(y, bins=6, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Values')
plt.show() 
```

## Histogram Analysis

The histogram illustrates how frequently different values occur in the dataset. Instead of focusing on individual observations, it focuses on the overall distribution of values.

The distribution shows that some values appear more often than others. Histograms are useful for understanding the spread of data, detecting concentration areas, and identifying whether the data is evenly distributed or skewed toward certain ranges.

# 3.Subplots

```python
import pandas as pd
df=pd.read_csv("students.csv")

fig,axes=plt.subplots(1,2,figsize=(10,4))
axes[0].hist(df['age'], bins=8, edgecolor='black')
axes[0].set_xlabel('age')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Histogram of Student Ages')
axes[1].bar(df['age'], df['score'])
axes[1].set_xlabel('age')
axes[1].set_ylabel('score')
axes[1].set_title('Bar Plot of Age vs Score')
plt.tight_layout()
plt.show()
```

# Student Dataset Visualizations

This figure uses subplots to display multiple visualizations within the same window, making it easier to compare different aspects of the dataset simultaneously.

The first subplot is a histogram that illustrates the distribution of student ages. It helps identify the most common age groups and shows how ages are spread across the dataset.

The second subplot is a bar chart that compares students' math scores across different ages. This visualization highlights variations in academic performance and makes it easier to observe potential relationships between age and achievement.

Using subplots improves data analysis by presenting multiple perspectives of the same dataset in a clear and organized manner, allowing comparisons without creating separate figures.
