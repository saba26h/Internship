# 1.Pandas Library

```python
import pandas as pd

df=pd.read_csv("students.csv")
print(df.head())
print(df.info())
df.isnull().sum()

df.fillna({'score': df['score'].mean()}, inplace=True)
df.fillna({'hours_studied': df['hours_studied'].min()}, inplace=True)
```

## Data Loading and Preprocessing

In this section, the dataset is loaded using the Pandas library and inspected to understand its overall structure. The first few rows are displayed to verify that the data has been loaded correctly, while the dataset information is used to check the number of records, data types, and missing values.

Missing values are identified and handled using the `fillna()` function to ensure that the dataset is complete before performing any analysis. Cleaning the data is an important preprocessing step because it improves the reliability of the results and prevents errors during visualization and statistical calculations.

Finally, descriptive statistics are generated to summarize the dataset. These statistics provide information such as the mean, minimum, maximum, and distribution of numerical variables. This initial exploration helps us understand the characteristics of the data before moving on to deeper analysis and visualization.

# 2.Numpy Library

```python
import numpy as np

average_score = df['score'].mean()
print(f"Average Score: {average_score}")
max_age = df['age'].max()
print(f"Maximum Age: {max_age}")
```

## Numerical Analysis Using NumPy

NumPy is used to perform efficient numerical operations on the dataset. Several statistical measures are calculated, including the average score, maximum age, minimum score, and standard deviation.

The average score represents the overall academic performance of the students, while the maximum age identifies the oldest student in the dataset. The minimum score highlights the lowest academic performance, and the standard deviation measures how much the students' scores vary from the average.

These calculations provide valuable insights into the dataset by summarizing the main characteristics of the numerical values. Understanding these statistics helps identify trends, detect unusual values, and prepare the data for visualization and further analysis.

# 3.Matplotlib Library

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1,3,figsize=(15,5))

ax[0].hist(df['score'], bins=10, color='blue', alpha=0.7)
ax[0].set_title('Score Distribution')
ax[0].set_xlabel('Score')
ax[0].set_ylabel('Frequency')

ax[1].scatter(df['hours_studied'], df['score'], color='green', alpha=0.7)
ax[1].set_title('Hours Studied vs Score')
ax[1].set_xlabel('Hours Studied')
ax[1].set_ylabel('Score')

ax[2].bar(df['department'], df['score'], color='orange', alpha=0.7)
ax[2].set_title('Average Score by Department')
ax[2].set_xlabel('Department')
ax[2].set_ylabel('Average Score')

plt.tight_layout()
plt.show()
```

## Data Visualization and Interpretation

Three different visualizations are created to better understand the dataset.

### Histogram of Student Scores

The histogram illustrates how student scores are distributed across different score ranges. It reveals whether most students achieved similar scores or whether the scores are widely spread. A concentration of bars around a specific range indicates where the majority of students performed, while very low or very high scores can be identified as less frequent observations.

### Scatter Plot of Age vs Score

The scatter plot displays the relationship between students' ages and their scores. Each point represents one student. If the points form an upward trend, it suggests that scores tend to increase with age. If there is no clear pattern, it indicates that age may not have a strong influence on academic performance. Scatter plots are useful for identifying correlations, clusters, and possible outliers.

### Bar Chart of Average Scores

The bar chart provides a clear comparison of the average score. It allows readers to quickly interpret the central tendency of the dataset and visually compare this value with other statistics if additional categories are included. Bar charts are effective because they present numerical information in a simple and easy-to-understand format.

Overall, these visualizations transform raw numerical data into graphical representations that make trends, patterns, and relationships much easier to interpret than tables of numbers alone.
